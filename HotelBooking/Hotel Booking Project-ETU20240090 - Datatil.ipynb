```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#read hotel booking csv file
```

```python
df = pd.read_csv("hotel_bookings.csv")
```

## 1. What are the general booking trends?
###     How many bookings are made each month?

```python
#定義月份的順序
month_order = ['January', 'February', 'March', 'April', 'May', 'June', 
               'July', 'August', 'September', 'October', 'November', 'December']
#use pd.Categorical to set up sorting of months 
df['arrival_date_month'] = pd.Categorical(df['arrival_date_month'], categories=month_order, ordered=True)

bookings_per_month = df['arrival_date_month'].value_counts().sort_index()
print(bookings_per_month)
```

### Which hotel types (city or resort) receive the most bookings?

```python
most_bookings_HotelTypes = df['hotel'].value_counts()
print(most_bookings_HotelTypes)
print('The most boolings hotel types:')
City_hotel_count = most_bookings_HotelTypes['City Hotel']
Resort_hotel_count = most_bookings_HotelTypes['Resort Hotel']
if City_hotel_count > Resort_hotel_count:
    print('City hotel')
elif Resort_hotel_count > City_hotel_count:
    print('Resort hotel')
```

### What is the average lead time for bookings?

```python
average_lead_time = df['lead_time'].mean()
print(average_lead_time)
```

## 2. What are the cancellation trends?
###  What is the overall cancellation rate? 

```python
cancellation_rate = df['is_canceled'].mean() * 100
print(cancellation_rate,'%')
```

### Is there a difference in cancellation rates between city hotels and resort hotels? 

```python
cancellation_rate_ByHotel = df.groupby('hotel')['is_canceled'].mean() * 100
print(cancellation_rate_ByHotel)
```

### Is there a relationship between lead time and cancellation rates?

```python
lead_time_canceled = df[df['is_canceled'] == 1]['lead_time'].mean()
lead_time_not_canceled = df[df['is_canceled'] == 0]['lead_time'].mean()
print('Lead time and canceled:',lead_time_canceled)
print('Lead time and not canceled',lead_time_not_canceled)
```

## 3. What is the distribution of average daily revenue (ADR)?
### What is the average ADR for each type of hotel? 

```python
adr_by_hotel_type = df.groupby('hotel')['adr'].mean()
print(adr_by_hotel_type)
```

### Are there differences in ADR by season (month)? 

```python
adr_by_month = df.groupby('arrival_date_month')['adr'].mean()
print(adr_by_month)
```

### Which market segments generate the highest revenue?

```python
adr_market_segments=df.groupby('market_segment')["adr"].mean().sort_values()
print(adr_market_segments)
```

##  What factors influence special requests?
### Which customer types (families, repeat guests, couples) make the most special requests? 

```python
special_request_By_customer = df.groupby('customer_type')['total_of_special_requests'].mean()
print(special_request_By_customer)
```

### Is there a relationship between special requests and the cancellation rate? 

```python
special_requests_vs_cancel = df.groupby('total_of_special_requests')['is_canceled'].mean() * 100
print(special_requests_vs_cancel)
```
